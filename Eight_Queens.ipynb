# Eight Queens Problem

The Eight Queens Problem is a classic constraint satisfaction problem in which eight queens must be placed on an 8×8 chessboard so that no two queens attack each other.

Queens can attack each other if they are in the same row, the same column, or along the same diagonal. This project implements a local search algorithm based on the **Hill Climbing** approach with random restarts to find a valid solution.

## Solver Implementation

The board is represented using a list in which each index represents a column and its corresponding value represents the row containing the queen in that column.

The algorithm works as follows:

- **Random Initialization:** Queens are randomly placed on the board, with one queen in each column.
- **Conflict Detection:** The number of pairs of queens attacking each other is calculated.
- **Best Neighbor Selection:** The algorithm examines possible positions for each queen and selects a move that reduces the number of conflicts.
- **Hill Climbing:** The algorithm continues moving queens toward positions with fewer conflicts until no better neighboring state is available.
- **Random Restart:** If the algorithm reaches a state with conflicts and cannot improve it, the board is randomized and the search starts again.

The process continues until a solution with zero attacking pairs is found.

```python
import random


def isAttacking(ax, ay, bx, by):
    # Check if queen (ax, ay) is attacking queen (bx, by)
    return ax == bx or ay == by or abs(bx - ax) == abs(by - ay)


class Board:
    def __init__(self, n):
        self.size = n
        self.queens = [0] * n

    def randomize(self):
        for i in range(self.size):
            self.queens[i] = random.randint(0, self.size - 1)

    def threatendQueens(self):
        ctr = 0

        for i in range(self.size - 1):
            for j in range(i + 1, self.size):
                if isAttacking(i, self.queens[i], j, self.queens[j]):
                    ctr += 1

        return ctr

    def transToBestNeighbor(self):
        min = self.threatendQueens()
        chosenQueen = -1
        moveTo = -1

        for i in range(self.size):
            for j in range(self.size):
                before = self.queens[i]
                self.queens[i] = j

                if min > self.threatendQueens():
                    min = self.threatendQueens()
                    chosenQueen = i
                    moveTo = j

                self.queens[i] = before

        if chosenQueen != -1:
            self.queens[chosenQueen] = moveTo
            return True
        else:
            return False

    def solve(self):
        while True:
            while self.transToBestNeighbor():
                pass

            if self.threatendQueens() == 0:
                return True

            self.randomize()

    def printBoard(self):
        for i in self.queens:
            print(i, end=' ')

        print('\n')


# Create an 8-Queens puzzle
puzzle = Board(8)

# Solve the puzzle
puzzle.solve()

# Display the solution
print('Solution:')
puzzle.printBoard()

# Display the number of attacking pairs
print('Number of attacking pairs:', puzzle.threatendQueens())
```
