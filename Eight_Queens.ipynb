# Eight Queens Problem

The Eight Queens Problem is a classical Artificial Intelligence problem in which eight queens must be placed on an 8×8 chessboard so that no two queens attack each other. Since queens can attack horizontally, vertically, and diagonally, the objective is to find a configuration with no conflicts.

This project uses **Hill Climbing with Random Restart** to find a valid solution through iterative local search.

## Board Representation and Conflict Detection

The board is represented by a one-dimensional list. Each index represents a column, and its value represents the row of the queen in that column. This ensures that each column contains exactly one queen.

The `isAttacking()` function checks whether two queens share a row, column, or diagonal. The `threatendQueens()` method counts all attacking pairs and serves as the evaluation function. The goal is to reduce this value to zero.

## Search Algorithm

The algorithm begins with a random board configuration. At each step, `transToBestNeighbor()` examines possible row positions for every queen and selects a move that reduces the number of conflicts.

The search continues while an improving move exists. If no better neighboring state is found and conflicts remain, the algorithm performs a **Random Restart** by generating a new random configuration. This process repeats until a solution with zero conflicts is reached.

## Implementation and Execution

The `Board` class implements board initialization, randomization, conflict counting, neighbor selection, and the solving process. Finally, an 8-Queens puzzle is created, solved, and its solution is displayed.

```python
import random

def isAttacking(ax,ay,bx,by): # check if queen (ax,ay) is attacking queen (bx,by)
    return ax==bx or ay==by or abs(bx-ax)==abs(by-ay)

class Board:
    def __init__(self,n):
        self.size = n
        self.queens = [0]*n

    def randomize(self):
        for i in range(self.size):
            self.queens[i] = random.randint(0,self.size-1)

    def threatendQueens(self):
        ctr = 0
        for i in range(self.size-1):
            for j in range(i+1,self.size):
                if isAttacking(i,self.queens[i],j,self.queens[j]):
                    ctr+=1
        return ctr

    def transToBestNeighbor(self):
        min = self.threatendQueens()
        chosenQueen = -1
        moveTo = -1

        for i in range(self.size):
            for j in range(self.size):
                before = self.queens[i]
                self.queens[i] = j

                if min > self.threatendQueens():
                    min = self.threatendQueens()
                    chosenQueen = i
                    moveTo = j

                self.queens[i] = before

        if chosenQueen!=-1:
            self.queens[chosenQueen] = moveTo
            return True
        else:
            return False

    def solve(self):
        while True:
            while self.transToBestNeighbor():
                pass

            if self.threatendQueens()==0:
                return True

            self.randomize()

    def printBoard(self):
        for i in self.queens:
            print(i,end=' ')
        print('\n')

# Create and solve the Eight Queens puzzle
puzzle = Board(8)
puzzle.solve()

# Display the solution
print('Solution:')
puzzle.printBoard()
print('Number of attacking pairs:', puzzle.threatendQueens())
```
