# Eight Queens Problem

## Introduction

The Eight Queens Problem is a classical problem in artificial intelligence and combinatorial optimization. The objective is to place eight queens on an 8×8 chessboard in such a way that no two queens attack each other.

In chess, a queen can move horizontally, vertically, and diagonally. Therefore, a valid arrangement must satisfy three constraints:

- No two queens can be placed in the same row.
- No two queens can be placed in the same column.
- No two queens can be placed on the same diagonal.

This project solves the Eight Queens Problem using a local search approach based on Hill Climbing with Random Restart. Instead of exploring every possible board configuration, the algorithm starts with a random configuration and iteratively improves it by reducing the number of attacking pairs.

## Board Representation

The chessboard is represented using a one-dimensional list rather than an 8×8 matrix. This representation simplifies the search process because exactly one queen is assigned to each column.

The index of the list represents the column of a queen, while the value stored at that index represents its row position.

For example, the list `[0, 4, 7, 5, 2, 6, 1, 3]` represents a board configuration in which the queen in column 0 is placed in row 0, the queen in column 1 is placed in row 4, and so on.

This representation guarantees that no two queens occupy the same column. The algorithm therefore focuses on finding row assignments that minimize conflicts between queens.

## Detecting Attacking Queens

The `isAttacking()` function determines whether two queens can attack each other. It receives the coordinates of two queens and checks whether they share a row, a column, or a diagonal.

Two queens are attacking each other if at least one of the following conditions is satisfied:

1. Their column coordinates are equal.
2. Their row coordinates are equal.
3. The absolute difference between their column coordinates is equal to the absolute difference between their row coordinates. This condition identifies queens located on the same diagonal.

The `threatendQueens()` method uses this function to count all pairs of attacking queens on the board. A valid solution is obtained when this count becomes zero.

## Hill Climbing with Random Restart

The algorithm uses a local search strategy to find a configuration with no attacking queens.

### 1. Random Initialization

The `randomize()` method assigns a random row to each queen. Since each queen is associated with a different column, this creates an initial board configuration with exactly one queen in every column.

### 2. Evaluating the Current State

The `threatendQueens()` method calculates the number of pairs of queens that attack each other. This value is used as the evaluation function, where a lower number indicates a better board configuration. The goal is to minimize this value until it reaches zero.

### 3. Selecting the Best Neighbor

The `transToBestNeighbor()` method examines possible moves for each queen. It temporarily moves a queen to every possible row in its column and evaluates the resulting board configuration.

If a move reduces the number of attacking pairs, the method records that queen and its new row. After examining the possible moves, the best improving move is applied to the board.

If no move can reduce the number of conflicts, the method returns `False`, indicating that the algorithm has reached a local optimum.

### 4. Random Restart

Hill Climbing can become trapped in a local optimum, where no neighboring state provides an improvement even though the current configuration is not a valid solution.

To overcome this limitation, the `solve()` method uses Random Restart. Whenever the algorithm cannot improve the current configuration and the number of conflicts is still greater than zero, it randomly initializes the board again and starts a new search.

The process continues until a configuration with zero attacking pairs is found.

## Implementation

The `Board` class contains the main functionality required to represent and solve the problem.

- `__init__(n)`: Initializes the board with a specified size and creates a list to store queen positions.
- `randomize()`: Randomly assigns a row position to each queen.
- `threatendQueens()`: Counts the number of pairs of queens attacking each other.
- `transToBestNeighbor()`: Searches for a move that reduces the number of conflicts and applies the best improving move found.
- `solve()`: Repeatedly applies Hill Climbing and Random Restart until a valid solution is found.
- `printBoard()`: Displays the final queen positions.

The complete implementation and execution of the algorithm are provided in the following code cell.

```python
import random


def isAttacking(ax, ay, bx, by):
    # Check if queen (ax, ay) is attacking queen (bx, by)
    return ax == bx or ay == by or abs(bx - ax) == abs(by - ay)


class Board:
    def __init__(self, n):
        self.size = n
        self.queens = [0] * n

    def randomize(self):
        for i in range(self.size):
            self.queens[i] = random.randint(0, self.size - 1)

    def threatendQueens(self):
        ctr = 0
        for i in range(self.size - 1):
            for j in range(i + 1, self.size):
                if isAttacking(i, self.queens[i], j, self.queens[j]):
                    ctr += 1
        return ctr

    def transToBestNeighbor(self):
        min = self.threatendQueens()
        chosenQueen = -1
        moveTo = -1

        for i in range(self.size):
            for j in range(self.size):
                before = self.queens[i]
                self.queens[i] = j

                if min > self.threatendQueens():
                    min = self.threatendQueens()
                    chosenQueen = i
                    moveTo = j

                self.queens[i] = before

        if chosenQueen != -1:
            self.queens[chosenQueen] = moveTo
            return True
        else:
            return False

    def solve(self):
        while True:
            while self.transToBestNeighbor():
                pass

            if self.threatendQueens() == 0:
                return True

            self.randomize()

    def printBoard(self):
        for i in self.queens:
            print(i, end=' ')
        print('\n')


# Create an 8-Queens puzzle
puzzle = Board(8)

# Solve the puzzle
puzzle.solve()

# Display the solution
print('Solution:')
puzzle.printBoard()

# Display the number of attacking pairs
print('Number of attacking pairs:', puzzle.threatendQueens())
```
